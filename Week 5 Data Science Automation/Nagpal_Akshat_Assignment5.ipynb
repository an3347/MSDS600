## Optional Tasks

### Calculate prediction percentiles

```python
loaded_model = load_model('best_churn_model', verbose=False)
```

```python
# Get predictions on the training data
# train_predictions = predict_model(loaded_model, data=df)
df_to_predict = df[df.columns[:-1]]

train_predictions = loaded_model.predict_proba(df_to_predict)
```

```python
from scipy.stats import percentileofscore

train_predictions = loaded_model.predict_proba(df_to_predict)

# Extract churn probabilities using the correct column name
train_churn_probabilities = pd.DataFrame(train_predictions, columns=['Probability_0', 'Probability_1'])

# Calculate percentiles for the training data churn probabilities
train_percentiles = train_churn_probabilities.apply(lambda x: percentileofscore(train_churn_probabilities, x))
train_percentiles.columns = ['Percentile_0', 'Percentile_1']

# Store the training data churn probabilities and their percentiles
train_probability_distribution = pd.concat([train_churn_probabilities['Probability_1'], train_percentiles['Percentile_1']], axis=1)
train_probability_distribution.columns = ['Churn Probability', 'Probability Percentile']

display(train_probability_distribution.head())
display(train_probability_distribution.describe())
```

```python
train_probability_distribution_path = 'train_probability_distribution.csv'
train_probability_distribution.to_csv(train_probability_distribution_path, index=False)
print(f"Training probability distribution saved to {train_probability_distribution_path}")
```

### h2o

```python
import h2o
from h2o.frame import H2OFrame

# Initialize H2O cluster
h2o.init()
```

```python
# Load the churn data
h2o_df = h2o.import_file('data/churn_data.csv')

# Convert the target variable to a factor
h2o_df['Churn'] = h2o_df['Churn'].asfactor()

# Identify feature and target columns
target = 'Churn'
features = h2o_df.columns
features.remove(target)

print("H2OFrame loaded and target converted to factor.")
print("Features:", features)
print("Target:", target)
```

```python
from h2o.automl import H2OAutoML

# Initialize H2O AutoML
aml = H2OAutoML(max_runtime_secs=300, seed=42) # Set max_runtime_secs to 300 seconds (5 minutes)

# Train the AutoML model
aml.train(x=features, y=target, training_frame=h2o_df)

print("H2O AutoML training complete.")
```

```python
# Get the AutoML leaderboard
leaderboard = aml.leaderboard

print("H2O AutoML Leaderboard:")
leaderboard.head()
```

H2O’s stacked ensemble doutperforms PyCaret’s KNN on the churn dataset: the H2O model achieves an AUC of 0.8428 compared to KNN’s 0.7666. KNN excels in speed—training in just 0.014 seconds—making it ideal for quick prototyping, but it falls short on discrimination and probability calibration. H2O’s minutes-long training yields more reliable probability estimates and better handling of class imbalance, enabling threshold tuning, cost-sensitive learning, and robust deployment for retention campaigns.
